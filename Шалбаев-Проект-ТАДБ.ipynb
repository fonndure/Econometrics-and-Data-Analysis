# Численность населения в трудоспособном возрасте, имеющего уровень высшего проффесионального образования - бакалавр и магистр

 ## Первый файл - бакалавр

```python
# читаем csv-файл
with open('data-bachelor.csv', 'r', encoding='utf-8-sig') as file:
    for a in file:
        print(a.strip().split(','))
```

```python
# Часть 1: Чтение данных из файла и создание словаря
# Создаем пустой словарь, где будут храниться данные по регионам
dict1 = {}
# Открываем файл для чтения
with open('data-bachelor.csv', 'r', encoding='utf-8-sig') as file:
    # Построчно читаем весь файл
    for line in file:
        row = line.strip().split(';')
        if row[0] == 'Количество': # Если первый элемент строки = "Количество" - это заголовок
            head = row # Присваиваем переменной head - заголовок
            continue   # Переходим к следующей строке файла
        if len(row) == 3:  # Если в строке 3 элемента (столбца)
            region = row[1].strip() # Записываем в переменную второй столбец "Субъект"
            value_b = float(row[0]) # Записываем в переменную первый столбец "Количество"
            dict1[region] = {'2016': value_b} # Добавляем в словарь: ключ - название региона, значение - словарь с данными за 2016
# Выводим получившийся словарь
print(dict1)
print(dict1['Омская область'])
```

```python
# Часть 2: Расчет статистических показателей
# Создаем список со всеми значениями за 2016 год из словаря
values = [region1['2016'] for region1 in dict1.values()]
a = sum(values)       # считаем все значения               - Для подсчета среднего значения
b = len(values)       # считаем кол-во значений(регионов)  - Для подсчета среднего значения
average = a/b # Расчёт среднего значения

# Вычисляем среднеквадратическое отклонение:
# 1. Для каждого значения вычисляем квадрат разности со средним: (x - average) ** 2
# 2. Суммируем все эти квадраты
# 3. Делим сумму на количество значений
# 4. Извлекаем квадратный корень
stand = (sum((x - average) ** 2 for x in values) / len(values)) ** 0.5

sorted_val = sorted(values) # Сортируем значения по возрастанию для медианы
n = len(sorted_val)         # Получаем кол-во отсорт-ных знач.
mid = n // 2                # Находим индекс среднего элемента
print('Статистические показатели за 2016 год:')
print(f'Среднее значение = {round(average, 2)}')
print(f'Максимум = {max(values)}')
print(f'Минимум = {min(values)}')
print(f'Среднеквадратическое отклонение = {round(stand, 2)}')
print(f'Медиана = {sorted_val[mid]}')
```

## Рейтинг бакалавров

```python
# Часть 3: Создание рейтинга регионов
# Снова читаем файл и создаем словать (повтор как в первой часте)
dict1 = {}
with open('data-bachelor.csv', 'r', encoding='utf-8-sig') as file:
    for line in file:
        row = line.strip().split(';')
        if row[0] == 'Количество':
            head = row
            continue
        if len(row) == 3:
            region = row[1].strip()
            value_b = float(row[0])
            dict1[region] = {'2016': value_b}

print(dict1)
list1 = [(region, region1['2016']) for region, region1 in dict1.items()]  # Создаем список кортежей: (название региона, значение за 2016 год)
list1.sort(key=lambda x: x[1], reverse=True)  # Сортируем список по значениям (2-ой элем.кортежа) в порядке убывания

# Создаем рейтинг: добавляем порядковый номер к каждому элементу
# i + 1 - номер в рейтинге (начало с 1)
# enumerate дает нам индекс (i) и значение (region, value)
rating = [(i + 1, region, value) for i, (region, value) in enumerate(list1)]
print("Рейтинг регионов-2016:")         # Выводим название таблицы
print("Рейтинг | Регион | Значение")    # Выводим шапку таблицы

# Выводим каждый регион из рейтинга по индексу
for rank in rating:
    print(f"{rank[0]} | {rank[1]} | {rank[2]}")
```

## Новый csv-файл (бакалавры)

```python
# Часть 4: Сохранение рейтинга в csv-файл
# Открываем новый файл для записи
with open('regionsbachelor2016.csv', 'w', encoding='utf-8-sig') as file:
    file.write("Рейтинг;Регион;Значение 2016\n")   # Записываем заголовок в файл
    # Записываем данные по каждому региону
    for rank in rating:
        file.write(f"{rank[0]};{rank[1]};{rank[2]}\n") # Формируем строку с разделителем ;
    print("Так выглядит новый файл:")
# Открываем только что созданныый файл для чтения
with open('regionsbachelor2016.csv', 'r', encoding='utf-8-sig') as file:
    # Построчно выводим содержимое файла на экран
    for line in file:
        print(line.strip())
```

## Второй файл - магистр

```python
# Часть 1: Анализ данных магистров (аналогично бакалаврам)
# Создам новый пустой словарь для хранения данных по магистрам
dict2 = {}
# Открываем файл с данными магистров для чтения
# 'r' - read
# encoding='utf-8-sig' - чтобы русские буквы читались праильно
with open('data-master.csv', 'r', encoding='utf-8-sig') as file:
    # Начинаем читать файл построчно
    for line in file:
        row = line.strip().split(';') # Убираем пробелы в начале и конце строки и делим на части по ";"
        if row[0] == 'Количество': # Определяем шапку таблицы
            head = row             # Снова присваиваем
            continue               # Пропуск этой строки, без обработки её данных
        if len(row) == 3:   # Если в строке 3 столбца
            region = row[1].strip()  # То присваимвем переменной второй столбец - Субъекты
            value_b = float(row[0])  # То присваиваем переменной первый столбец - Значения, превращаем их в десятичное число
            dict2[region] = {'2016': value_b} # Добавляем в словарь: ключ - название региона, значение - словарь с данными за 2016 год
# Выводим получившийся словарь
print(dict2)
```

```python
# Часть 2: Расчет статистических показателей
# Создаем список со всеми значениями за 2016 год из словаря
values = [region1['2016'] for region1 in dict2.values()]
a = sum(values)         # считаем все значения - Для подсчета среднего значения
b = len(values)         # считаем кол-во значений в списке - Для подсчета среднего значения
average = a/b           # Расчёт среднего арифметического значения: сумма / кол-во

# Считаем среднеквадратическое отклонение:
# 1. Для каждого значения вычисляем квадрат разности со средним: (x - average) ** 2
# 2. Суммируем все эти квадраты
# 3. Делим сумму на количество значений
# 4. Извлекаем квадратный корень
stand = (sum((x - average) ** 2 for x in values) / len(values)) ** 0.5

sorted_val = sorted(values) # Сортируем список значений по возраст.
n = len(sorted_val)         # Получаем кол-во отсорт-ных знач.
mid = n // 2                # Находим индекс среднего элемента
print('Статистические показатели за 2016 год:')
print(f'Среднее значение = {round(average, 2)}')
print(f'Максимум = {max(values)}')
print(f'Минимум = {min(values)}')
print(f'Среднеквадратическое отклонение = {round(stand, 2)}')
print(f'Медиана = {sorted_val[mid]}')
```

## Рейтинг магистров
